## Exercise 2: Select links

Extract each anchor's visible text and href into a list of dictionaries. Ignore anchors without an href.

````python
html = '<a href="/home">Home</a><a href="/about">About</a><a>Missing</a>'

# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html, 'html.parser')
links = [{'text': link.get_text(strip=True), 'href': link['href']}
         for link in soup.select('a[href]')]
print(links)
```

## Exercise 3: Scrape a table

Parse the table and create a list of dictionaries with keys name and score. Expected records: Ana/8 and Ben/10.

````python
html = '<table><tr><th>Name</th><th>Score</th></tr><tr><td>Ana</td><td>8</td></tr><tr><td>Ben</td><td>10</td></tr></table>'

# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html, 'html.parser')
rows = soup.select('tr')[1:]
records = []
for row in rows:
    name, score = [cell.get_text(strip=True) for cell in row.select('td')]
    records.append({'name': name, 'score': int(score)})
print(records)
```

</details>

## Exercise 4: Clean extracted text

Extract the article text, normalize its whitespace, and save title and body to a dictionary. Use get_text() or an equivalent BeautifulSoup method.

````python
html = '<article><h2>News</h2><p>  Python  is useful. </p></article>'

# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html, 'html.parser')
article = soup.select_one('article')
title = article.select_one('h2').get_text(strip=True)
body = ' '.join(article.select_one('p').get_text().split())
print({'title': title, 'body': body})
```

</details>

```python
import requests
from bs4 import BeautifulSoup

try:
    response = requests.get('https://example.com', timeout=5)
    response.raise_for_status()
    soup = BeautifulSoup(response.text, 'html.parser')
    print(response.status_code, soup.title.get_text(strip=True))
except requests.exceptions.Timeout:
    print('The request timed out.')
except requests.exceptions.RequestException as error:
    print('Request failed:', error)
```
