## Exercise 4: Compare and combine

````python
for age in [17, 18, 64, 65]:
    at_least_18 = age >= 18
    under_65 = age < 65
    within_limits = at_least_18 and under_65
    print(age, at_least_18, under_65, within_limits)
````
