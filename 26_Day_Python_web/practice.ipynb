## Exercise 3: Query parameters

Add a GET /square route that reads an integer x from the query string and returns its square. Return a 400 response for missing or invalid x. Test x=4 and x=bad.

````python
# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python
from flask import Flask, request

app = Flask(__name__)

@app.get('/square')
def square():
    try:
        number = int(request.args['x'])
    except (KeyError, ValueError):
        return 'Invalid x', 400
    return str(number ** 2)

client = app.test_client()
print(client.get('/square?x=4').get_data(as_text=True))
print(client.get('/square?x=bad').status_code)
```

</details>

## Exercise 4: Form handling

Add a POST /analyze route that reads text from form data and returns its word count. Use the test client to submit 'Python is fun'; expected count: 3.

````python
# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python
from flask import Flask, request

app = Flask(__name__)

@app.post('/analyze')
def analyze():
    text = request.form.get('text', '')
    return str(len(text.split()))

client = app.test_client()
response = client.post('/analyze', data={'text': 'Python is fun'})
print(response.status_code, response.get_data(as_text=True))
```

Render a small HTML template with a page title and a list of three topics. You may use render_template_string in the notebook. Inspect the rendered HTML with the test client.

<details>

```python
from flask import Flask, render_template_string

app = Flask(__name__)

@app.get('/topics')
def topics():
    template = '<h1>{{ title }}</h1><ul>{% for topic in topics %}<li>{{ topic }}</li>{% endfor %}</ul>'
    return render_template_string(template, title='Topics',
                                  topics=['Python', 'Flask', 'Pandas'])

client = app.test_client()
print(client.get('/topics').get_data(as_text=True))
```
