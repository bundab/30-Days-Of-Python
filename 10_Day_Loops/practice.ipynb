## Exercise 1: For and range

Print the even numbers from 2 through 20, then calculate their total. Expected total: 110.

````python
# TODO: Write your solution here.

limit = 20
index = 1
total = 0

# Solution A
while index <= limit:
    if index % 2 == 0:
        print(index)
        total = total + index
    index += 1
print('The expected total value is', total)

# Solution B
total = 0
for number in range(2, 21, 2):
    print(number)
    total += number
print('The expected total value is', total)
````

## Exercise 3: Skip and stop

Loop over the values below. Skip negative values with continue, stop at the first zero with break, and print only values before that point.

````python
values = [4, -2, 7, 0, 9]

# TODO: Write your solution here.

for value in values:
    if value < 0:
        print('Negative value:', value)
        continue
    elif value == 0:
        break
    print('Positive value:', value)
````

## Exercise 4: Count letters

Use a loop and a dictionary to count the letters in 'banana'. Expected counts: b=1, a=3, n=2.

````python
word = 'banana'

# TODO: Write your solution here.

letters = {}
for letter in word:
    if letter not in letters and letter.isalpha():
        letters[letter] = 1
    elif letter in letters:
        letters[letter] += 1

print(letters)
````

## Exercise 5: Nested multiplication table

Use nested loops to print a 3 by 3 multiplication table for 1 through 3. Each row should contain three products.

````python
# TODO: Write your solution here.

matrix = list()
# product = 0

for index_x in range(0,3):
    column = list()
    for index_y in range(0,3):
        if index_x > index_y:
            column.append(index_x)
        else:
            column.append(index_y)
    matrix.append(column)

print(matrix.__getitem__(0))
print(matrix.__getitem__(1))
print(matrix.__getitem__(2))
````
