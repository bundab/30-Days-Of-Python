```python
import importlib

# AttributeError: the object has no such attribute or method.
try:
    (3).append(1)
except AttributeError as error:
    print(type(error).__name__)

# ModuleNotFoundError: the named module cannot be imported.
try:
    importlib.import_module('package_that_does_not_exist')
except ModuleNotFoundError as error:
    print(type(error).__name__)

# KeyError: the dictionary has no such key.
try:
    {'name': 'Ada'}['age']
except KeyError as error:
    print(type(error).__name__)
```
