````python
a = 10
b = 2

# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python
for b in (2, 0):
    try:
        result = a / b
    except ZeroDivisionError:
        print('cannot divide by zero')
    else:
        print(result)
    finally:
        print('finished')
```
