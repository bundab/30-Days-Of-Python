## Exercise 3: Names and lengths

Build a full_name variable from first_name and last_name. Print its length and whether first_name has more characters than last_name.

````python
first_name = 'Ada'
last_name = 'Lovelace'

# TODO: Write your solution here.
function_full_name = lambda first_name, last_name: first_name + ' ' + last_name
function_which_one_is_longer = lambda first_name, last_name: first_name if len(first_name) > len(last_name) else last_name if len(last_name) > len(first_name) else first_name + ' and ' + last_name + ' are the same length, so none of them is'

full_name = function_full_name(first_name, last_name)
print(full_name, 'is the full name of the person.\tLength of the full name:', len(full_name))

which_one_is_longer = function_which_one_is_longer(first_name, last_name)
print(which_one_is_longer, 'is the longer name.\tLength of the longer name:', len(which_one_is_longer))
````

## Exercise 5: Built-in toolkit

Given the list below, use len, min, max, sum, and abs to report its size, smallest value, largest value, total, and absolute value of the smallest number.

````python
measurements = [4, -7, 10, 2]

length = len(measurements)
minimum = min(measurements)
maximum = max(measurements)
total = sum(measurements)
absolute_values = [abs(item) for item in measurements]

print('Length:', length)
print('Minimum:', minimum)
print('Maximum:', maximum)
print('Sum:', total)
print('Absolute values:', absolute_values)
````
