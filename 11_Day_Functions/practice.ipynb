## Exercise 4: Keyword arguments

Define describe_product(name, price, currency='EUR') and return a readable label. Call it once positionally and once using keyword arguments in a different order.

````python
def describe_product(name, price, currency='EUR'):
    return f'{name}: {price:.2f} {currency}'

print(describe_product('Pen', 2.5))
print(describe_product(price=3, name='Notebook', currency='USD'))
````

## Exercise 5: Function as an argument

Define apply_operation(a, b, operation) that returns operation(a, b). Pass separate add and multiply functions to it; check results 9 and 20 for inputs 4 and 5.

````python
def add(a, b):
    return a + b    

def multiply(a, b):
    return a * b

def apply_operation(a, b, operation):
    return operation(a, b)

# TODO: Write your solution here.

print(apply_operation(7, 65, add))
print(apply_operation(3, 86, multiply))
````
