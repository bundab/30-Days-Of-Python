## Exercise 2: Extract dates

Use re.findall to extract all dates in YYYY-MM-DD form. Expected: ['2024-01-05', '2025-11-20'].

````python
text = 'Start 2024-01-05; end 2025-11-20; note 05/01/2024.'

# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python

dates = re.findall(r'\b\d{4}-\d{2}-\d{2}\b', text)
print(dates)
```

## Exercise 3: Validate an identifier

Write a regex that matches an entire Python-style identifier beginning with a letter or underscore and followed by letters, digits, or underscores. Test 'user_2', '_x', '2users', and 'bad-name'.

````python
# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python
import re

pattern = r'[A-Za-z_][A-Za-z_0-9]*'
for candidate in ('user_2', '_x', '2users', 'bad-name'):
    print(candidate, bool(re.fullmatch(pattern, candidate)))
```

</details>

## Exercise 4: Normalize whitespace

Use re.sub to replace runs of spaces, tabs, or newlines with a single space, then strip outer whitespace. Expected result: 'Python is fun'.

````python
text = '  Python\t is  \n fun  '

# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python

cleaned = re.sub(r'\s+', ' ', text).strip()
print(cleaned)
```
