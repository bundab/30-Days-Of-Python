## Exercise 3: Remove and count

Remove one 'blue', pop the final item into a variable, then count the remaining 'blue' items. Print the list, popped item, and count.

````python
colors = ['red', 'blue', 'green', 'blue', 'yellow']

# TODO: Write your solution here.

colors.remove('blue')
last_item = colors.pop()
colors.insert(1, 'blue')
count_of_blue = colors.count('blue')

print('The final list looks something like this one: %s\nThe popped item is here: %s\nThe count number of the blue color in our list is: %d' %(colors, last_item, count_of_blue))
````

## Exercise 4: Copy versus alias

Make both an alias and an independent copy of the list. Change the first element in the original, then print all three lists and explain the outcome in a comment.

````python
numbers = [1, 2, 3]

# TODO: Write your solution here.

alias = numbers
alias2 = alias
numbers_copy = numbers.copy()
numbers[0] = 4
print("The original values: %s\nThe alias\' value: %s\nThe values of the copied list: %s\nThe alias2 looks like this: %s" %(numbers, alias, numbers_copy, alias2))
````

## Exercise 5: Combine and sort

Join the two lists, sort the joined list in ascending order, and create a reversed copy without modifying the sorted list. Expected sorted values: [1, 2, 3, 4, 5, 6].

````python
left = [5, 1, 3]
right = [4, 2, 6]

# TODO: Write your solution here.

joined_list = left + right
joined_list.sort()
reversed_joined_list = joined_list.copy()
reversed_joined_list.sort(reverse=True)

print('The original lists are %s and %s.\nThe joined list looks like this %s and shouldn\'t be the same as the reversed one.\nThe reversed joined list looks like this: %s' %(left, right, joined_list, reversed_joined_list))
````

<details>
<summary>Show example solution</summary>

```python
joined = left + right
joined.sort()
reversed_copy = joined[::-1]
print('sorted:', joined)
print('reversed copy:', reversed_copy)
```
