## Exercise 2: Math tools

Using math, calculate the ceiling of 4.2, the square root of 81, and the area of a circle with radius 2 using math.pi.

````python
# TODO: Write your solution here.

from math import ceil, sqrt, pow, pi

ceil = ceil(4.2)
square = sqrt(81)
area_of_circle = pow(2,2) * pi

print(f'The ceiling of 4.2 is: {ceil:.2f}\nThe square root of 81 is: {square:.0f}\nThe area of a circle with radius 2 is: {area_of_circle:.4f}')
````

## Exercise 3: Statistics tools

Using statistics, calculate the mean, median, and mode of the sample below.

````python
sample = [3, 4, 4, 6, 8]

# TODO: Write your solution here.

from statistics import *

mean = mean(sample)
median = median(sample)
mode = mode(sample)

print(f'This is the mean of the sample: {mean:.0f}\n'
      f'This is the median of the sample: {median:.0f}\n'
      f'This is the mide of the sample: {mode:.0f}')

````

## Exercise 4: Reproducible randomness

Create a local random.Random instance with seed 42. Generate five integers from 1 through 6. Run the cell twice and confirm the same sequence appears.

````python
# TODO: Write your solution here.

from random import Random

rng = Random(42)
randomized_number = rng.random()
randomized_integers = [rng.randint(1,6) for index in range(0, 5)]

print(f'This is the randomized number: {randomized_number:.4f}\n'
      f'This is the list of the fully randomized integers: {randomized_integers}')
````
