## Exercise 4: Inspect keys and values

Print a list of keys, a list of values, and a list of key-value pairs. Check whether 'price' is a key.

````python
product = {'name': 'pen', 'price': 2.5, 'stock': 14}

# TODO: Write your solution here.

print(product.items())

copy = product.copy()
print(copy.keys())
print(copy.items())

product.clear()
print(product)
del copy
````

## Exercise 5: Nested data

Read the city from the nested address, change the postal code to '1001', and add 'email': 'leo@example.com'. Print the final record.

````python
person = {'name': 'Leo', 'address': {'city': 'Oslo', 'postal_code': '1000'}}

# TODO: Write your solution here.
city = person.get('address').get('city')
print('This is the person\'s city: %s' %city)

person['address']['postal_code'] = '1001'
person['email'] = 'leo@example.com'

print('This is the final record of the dictionary called person: %s' %person)
````
