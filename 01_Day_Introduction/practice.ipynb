````python
number_one = 5
number_two = 17

print("addition:", number_one + number_two)
print("subtraction:", number_one - number_two)
print("multiplication:", number_one * number_two)
print("true division:", number_one / number_two)
print("floor division:", number_one // number_two)
print("remainder:", number_one % number_two)
print("exponentiation:", number_one ** number_two)
````

## Exercise 4: Container differences

Create a list, tuple, and set from the values 2, 2, 4, 6. Display each and its length. Write a short comment explaining the difference you observe.

````python
items = [1, 2, 3, 4, 5]

my_set = set(items)
my_tuple = tuple(items)
my_list = list(items)

for collection in (my_set, my_tuple, my_list):
    print("Collection:", collection, "Length:", len(collection))
````

<details>
<summary>Show example solution</summary>

```python

    print(collection, len(collection))
```

## Exercise 5: Distance challenge

Use arithmetic and exponentiation to calculate the Euclidean distance between (1, 2) and (7, 10). The expected result is 10.0. Do not import a distance library.

````python
point1 = (1,2)
point2 = (7, 10)

function = lambda p1, p2: ((p1[0] - p2[0]) ** 2 + (p1[1] - p2[1]) ** 2) ** 0.5

print("Distance between points:", function(point1, point2))
````
