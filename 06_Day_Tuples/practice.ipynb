## Exercise 3: Membership

Check whether 'Python' and 'Rust' are in the tuple. Print both Boolean results.

````python
skills = ('HTML', 'Python', 'SQL')

# TODO: Write your solution here.

contains_python = skills.__contains__('Python')
contains_rust = skills.__contains__('Rust')

print('The skills tuple contains the word Python: %s' %contains_python)
print('The skills tuple contains the word Rust: %s' %contains_rust)
````
