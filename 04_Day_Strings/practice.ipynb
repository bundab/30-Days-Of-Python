## Exercise 4: Clean and split

Strip outer spaces, convert to lowercase, replace 'coding' with 'Python', and split the resulting text into words. Expected words: ['i', 'enjoy', 'python', 'daily'].

````python
text = '  I enjoy coding daily  '

# TODO: Write your solution here.

text_without_whitespace = text.strip()
print('The original text is:', text, 'and the text without whitespace is:', text_without_whitespace)

replaced_text = text_without_whitespace.replace('coding', 'Python')
print('The text after replacement is:', replaced_text)

splitted_text = replaced_text.split(' ')
print('The splitted text is:', splitted_text)

print('Lowercased words:', [word.lower() for word in splitted_text])
````
