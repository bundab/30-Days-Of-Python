## Exercise 4: Inheritance

Define EBook as a subclass of Book with a file_size_mb attribute. Reuse the parent constructor and override describe() to include the file size.

````python
# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python
class EBook(Book):
    def __init__(self, title, author, file_size_mb):
        super().__init__(title, author)
        self.file_size_mb = file_size_mb

    def describe(self):
        return f'{super().describe()} ({self.file_size_mb} MB)'

ebook = EBook('Dune', 'Frank Herbert', 2.5)
print(ebook.describe())
```

## Exercise 5: Account challenge

Design a PersonAccount class with a name, a list of incomes, and a list of expenses. Implement add_income, add_expense, and balance. Test with incomes 100 and 50 and expense 40; expected balance: 110.

````python
# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python
class PersonAccount:
    def __init__(self, name):
        self.name = name
        self.incomes = []
        self.expenses = []

    def add_income(self, amount):
        self.incomes.append(amount)

    def add_expense(self, amount):
        self.expenses.append(amount)

    def balance(self):
        return sum(self.incomes) - sum(self.expenses)

account = PersonAccount('Mina')
account.add_income(100)
account.add_income(50)
account.add_expense(40)
print(account.balance())
```
