## Exercise 1: Transform numbers

Use one list comprehension to create the squares of 0 through 9. Expected first and last elements: 0 and 81.

````python
# TODO: Write your solution here.

from math import pow

squares = [pow(element, 2) for element in range(10)]

for index in range(10):
    print(f'The square of {index:.0f} is {squares[index]:.0f}.')
````

## Exercise 2: Filter and transform

Use one comprehension to keep positive even values and double them. Expected result: [4, 16, 20].

````python
numbers = [-4, 2, 3, 8, -1, 10]

# TODO: Write your solution here.

even_positive_numbers = [element * 2 for element in numbers if element % 2 == 0 and element > 0]
print(f'The positive and even numbers are the following in our list: {even_positive_numbers}')
````

## Exercise 3: Flatten a matrix

Flatten the matrix into a single list with a nested comprehension. Expected result: [1, 2, 3, 4, 5, 6].

````python
matrix = [[1, 2], [3, 4], [5, 6]]

# TODO: Write your solution here.

flattened_matrix = [element for row in matrix for element in row]
print(f'This is the flattened matrix that has been created: {flattened_matrix}.')
````

## Exercise 4: Comprehension with strings

Build a list of uppercase country names whose original names are longer than five characters.

````python
countries = ['Italy', 'Finland', 'Spain', 'Norway']

# TODO: Write your solution here.

uppercase_longer_countries = [country.upper() for country in countries if len(country) > 5]
print(f'These are the countries that are longer than five characters and also they are capitalized: {uppercase_longer_countries}')
````

## Exercise 5: Lambda practice

Create a lambda that accepts a number and returns its cube. Use it to build cubes of 1, 2, and 3. Then write a lambda factory that adds a chosen offset to its input.

````python
# TODO: Write your solution here.

from math import pow

numbers = (1, 2, 3)

cube = lambda number: pow(number, 3)
cubes_of_numbers = (cube(c) for c in numbers)

print(
    f'These are the cubes of the following numbers.\n'
    f'Numbers:\t{", ".join(f"{number:.0f}" for number in numbers)}\n'
    f'Cubes:\t\t{", ".join(f"{cube:.0f}" for cube in cubes_of_numbers)}.'
)


# I don't really understand the following part of the code...
# I guess something like this should work, but I'm not a 100% sure about that at all.
# Anyway, It is basically a lambda in a lambda, that can be used to implement different type of nested operations as short as possible.

f = lambda offset: lambda num: num + offset
set_offset = f(5)
add_offset_to_number = set_offset(7)

print(add_offset_to_number)
````
