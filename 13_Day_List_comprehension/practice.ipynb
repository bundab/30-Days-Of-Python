## Exercise 2: Filter and transform

Use one comprehension to keep positive even values and double them. Expected result: [4, 16, 20].

````python
numbers = [-4, 2, 3, 8, -1, 10]

# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python
doubled_positive_evens = [n * 2 for n in numbers if n > 0 and n % 2 == 0]
print(doubled_positive_evens)
```

## Exercise 3: Flatten a matrix

Flatten the matrix into a single list with a nested comprehension. Expected result: [1, 2, 3, 4, 5, 6].

````python
matrix = [[1, 2], [3, 4], [5, 6]]

# TODO: Write your solution here.
````

<summary>Show example solution</summary>

```python

```

## Exercise 4: Comprehension with strings

Build a list of uppercase country names whose original names are longer than five characters.

````python
countries = ['Italy', 'Finland', 'Spain', 'Norway']

# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python

print(selected)
```
