## Exercise 1: Map values

Use map and a named function or lambda to convert Celsius readings to Fahrenheit. Formula: F = C * 9/5 + 32.

````python
celsius = [0, 20, 100, 7]

# TODO: Write your solution here.

celsius_to_fahrenheit_converter = lambda a: a * 9 / 5 + 32
fahrenheit_values = tuple(map(celsius_to_fahrenheit_converter, celsius))

print(f'These are the temperatures in Celsius:\t{celsius}\n'
      f'These are the values in Fahrenheit:\t{fahrenheit_values}')
````

Use filter to keep names with at least five characters. Turn the result into a list.

````python
names = ['Ana', 'Benjamin', 'Cara', 'Derek']

# TODO: Write your solution here.

at_least_five_characters = lambda element: len(element) >= 5
filtered_names = tuple(filter(at_least_five_characters, names))

print(f'These are the filtered names of the "names" list: {filtered_names}.')
````

## Exercise 3: Reduce values

Use functools.reduce to multiply the integers 1 through 5. Expected result: 120.

````python
# TODO: Write your solution here.

from functools import reduce

numbers = [1, 2, 3, 4, 5]

def add_two_numbers(x, y):
    return x * y

multiplied_value = reduce(add_two_numbers, numbers)
print(f'This is the multiplied value of the given numbers: {numbers} and the value is: {multiplied_value}')
````

## Exercise 4: Closure

Write make_multiplier(factor) that returns a function multiplying its argument by factor. Verify that double(7) gives 14 and triple(7) gives 21.

````python
# TODO: Write your solution here.

def make_multiplier(factor):
    def multiply(number):
        return number * factor
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)

print(f'This is the doubled value:\t{double(5)}\n'
      f'This is the tripled value:\t{triple(6)}')
````

```python
from functools import wraps

def announce(function):
    @wraps(function)
    def wrapper(*args, **kwargs):
        print('Starting')
        result = function(*args, **kwargs)
        print('Finished')
        return result
    return wrapper

@announce
def add(a, b):
    return a + b

print(add(2, 3))
```
