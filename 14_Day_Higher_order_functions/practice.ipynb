## Exercise 5: Decorator

Write a decorator that prints 'Starting' before a function call and 'Finished' after it, while returning the wrapped function's result. Apply it to a function that adds two numbers.

````python
# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python
from functools import wraps

def announce(function):
    @wraps(function)
    def wrapper(*args, **kwargs):
        print('Starting')
        result = function(*args, **kwargs)
        print('Finished')
        return result
    return wrapper

@announce
def add(a, b):
    return a + b

print(add(2, 3))
```

</details>
