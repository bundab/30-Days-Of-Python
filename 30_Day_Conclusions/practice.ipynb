## Exercise 2: Validate records

Write a function valid_record(record) that accepts nonempty title, positive integer pages, and integer rating from 1 to 5. Use it to separate valid and invalid starter records. Expected: three valid and one invalid.

````python
# TODO: Write your solution here.
````

<details>
<summary>Show example solution</summary>

```python
def valid_record(record):
    title = record.get('title')
    pages = record.get('pages')
    rating = record.get('rating')
    return (isinstance(title, str) and bool(title.strip())
            and isinstance(pages, int) and not isinstance(pages, bool) and pages > 0
            and isinstance(rating, int) and not isinstance(rating, bool)
            and 1 <= rating <= 5)

valid = [record for record in records if valid_record(record)]
invalid = [record for record in records if not valid_record(record)]
print(len(valid), len(invalid))
```
