from importlib import metadata

examples = {'requests': 'requests.get', 'numpy': 'numpy.array',
            'pandas': 'pandas.DataFrame'}
for package, public_api in examples.items():
    try:
        distribution = metadata.distribution(package)
    except metadata.PackageNotFoundError:
        continue
    print(package, distribution.version)
    print(distribution.metadata.get('Summary'))
    print('Example API:', public_api)
    break
else:
    distribution = metadata.distribution('pip')
    print('pip', distribution.version)
    print(distribution.metadata.get('Summary'))
    print('Example to try after installing requests: requests.get')
```

from pathlib import Path

path = folder / 'practice_tools.py'
path.write_text('def is_even(number):\n    return number % 2 == 0\n', encoding='utf-8')
spec = importlib.util.spec_from_file_location('practice_tools', path)
practice_tools = importlib.util.module_from_spec(spec)
spec.loader.exec_module(practice_tools)
print(practice_tools.is_even(2), practice_tools.is_even(3))
```
